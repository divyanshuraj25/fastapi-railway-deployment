```python
%%writefile main.py
import os
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="AI Backend Service")

# Environment variables se config read karna
API_KEY = os.getenv("API_KEY", "default-fallback-key")
MODEL_NAME = os.getenv("MODEL_NAME", "gpt-4o-mini")
SIMILARITY_THRESHOLD = float(os.getenv("SIMILARITY_THRESHOLD", "0.75"))

class QueryRequest(BaseModel):
    query: str

@app.get("/")
def root():
    return {"status": "ok", "message": "Service is live"}

@app.get("/health")
def health_check():
    return {
        "status": "healthy",
        "model": MODEL_NAME,
        "threshold": SIMILARITY_THRESHOLD
    }

@app.post("/predict")
def predict(request: QueryRequest):
    return {
        "query": request.query,
        "model_used": MODEL_NAME,
        "score": SIMILARITY_THRESHOLD,
        "response": f"Processed: '{request.query}' successfully."
    }
```

```python
%%writefile requirements.txt
fastapi>=0.110.0
uvicorn[standard]>=0.29.0
pydantic>=2.0.0
requests>=2.31.0
```

```python
%%writefile Dockerfile
FROM python:3.11-slim

# Working directory
WORKDIR /app

# Prevent Python buffering
ENV PYTHONUNBUFFERED=1

# Dependencies install karna
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Application code copy karna
COPY . .

# Railway $PORT expose aur dynamic binding
EXPOSE 8000
CMD ["sh", "-c", "uvicorn main:app --host 0.0.0.0 --port ${PORT:-8000}"]
```
